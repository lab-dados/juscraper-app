## 2. Fazer a busca

`scraper.cjsg(...)` busca jurisprudencia de 2o grau; `scraper.cjpg(...)` busca o banco de sentencas de 1o grau (disponivel em TJES, TJSP e TJTO).

O argumento `paginas` aceita um inteiro (baixa da pagina 1 ate N), um `range`, ou `None` (todas). Os demais filtros sao os mesmos da ferramenta web. Cole aqui o codigo da aba **Codigo** ou edite o exemplo:

```python
import juscraper as jus

scraper = jus.scraper("tjsp")
df = scraper.cjsg(
    "dano moral",
    paginas=range(1, 6),  # paginas 1 a 5; aumente conforme precisar
    tipo_decisao="acordao",
)
print(f"{len(df)} linhas")
df.head()
```

## 3. Salvar os resultados

```python
df.to_csv("resultado.csv", index=False)
df.to_excel("resultado.xlsx", index=False)  # requer openpyxl: %pip install -q openpyxl

from google.colab import files  # so no Colab
files.download("resultado.csv")
```
